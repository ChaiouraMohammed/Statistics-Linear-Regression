```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#Hours of study X and corresponding scores Y
X = np.array([
    2.5, 3.0, 3.5, 4.0, 4.5,
    5.0, 5.5, 6.0, 6.5, 7.0,
    7.5, 8.0, 8.5, 9.0, 9.5,
    10.0, 10.5, 11.0, 11.5, 12.0,
    12.5, 13.0, 13.5, 14.0, 14.5,
    15.0, 15.5, 16.0, 16.5, 17.0
])
Y = np.array([
    48, 52, 55, 58, 61,
    63, 65, 67, 70, 72,
    73, 75, 77, 79, 80,
    82, 83, 84, 86, 87,
    88, 89, 90, 91, 92,
    93, 94, 94, 95, 96
])
```

```python
xMean = np.mean(X)
yMean = np.mean(Y)
print("Mean of X:", xMean)
print("Mean of Y:", yMean)
```

```python
beta1 = np.sum((X - xMean) * (Y - yMean)) / np.sum((X - xMean) ** 2)
beta0 = yMean - beta1 * xMean
print("Slope beta1:", beta1)
print("Intercept beta0:", beta0)
```

```python
yPred = beta0 + beta1 * X
```

```python
for hours, actual, predicted in zip(X, Y, yPred):
    print(
        f"Study Hours: {hours:.1f} | "
        f"Actual Score: {actual} | "
        f"Predicted Score: {predicted:.2f}"
    )
```

```python
results = pd.DataFrame({
    "Study Hours": X,
    "Actual Score": Y,
    "Predicted Score": yPred
})
```

```python
actualScoreMean = np.mean(Y)
predictredScoreMean = np.mean(yPred)
print("Mean of Actual Scores:", actualScoreMean)
print("Mean of Predicted Scores:", predictredScoreMean)
```

```python
sse = np.sum((Y - yPred) ** 2)
sst = np.sum((Y - yMean) ** 2)
rSquared = 1 - (sse / sst)
print("R-squared:", rSquared)
print("Sum of Squares due to Error (SSE):", sse)
print("Total Sum of Squares (SST):", sst)
```

```python
plt.scatter(X, Y, color='blue', label='Actual Scores')
plt.plot(X, yPred, color='red', label='Regression Line')
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(X,Y,label="Observed exam scores")
plt.plot(X,yPred,label="Least-squares regression line")
plt.xlabel("Weekly Stud-y Hours")
plt.ylabel("Exam Score")
plt.title("Relationship Between Study Hours and Exam Scores")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
